```python
# ==========================================
# STUDENT DATA ANALYSIS USING PANDAS
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# ------------------------------------------
# 1. Upload CSV file
# ------------------------------------------

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("File uploaded:", file_name)


# ------------------------------------------
# 2. Load CSV file
# ------------------------------------------

df = pd.read_csv(file_name, sep="|")

# Clean column names
df.columns = df.columns.str.strip().str.strip("|").str.strip()

# Remove empty columns
df = df.dropna(axis=1, how="all")


# ------------------------------------------
# 3. Convert numerical columns
# ------------------------------------------

df["Student_ID"] = pd.to_numeric(df["Student_ID"], errors="coerce")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
df["Attendance"] = pd.to_numeric(df["Attendance"], errors="coerce")


# ------------------------------------------
# 4. Display dataset
# ------------------------------------------

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ------------------------------------------
# 5. Basic Statistics
# ------------------------------------------

print("\nBasic Statistics:")
display(df.describe())


# ------------------------------------------
# 6. Calculate Average Age
# ------------------------------------------

average_age = df["Age"].mean()

print("\nAverage Age:", round(average_age, 2))


# ------------------------------------------
# 7. Calculate Average Marks
# ------------------------------------------

average_marks = df["Marks"].mean()

print("Average Marks:", round(average_marks, 2))


# ------------------------------------------
# 8. Calculate Average Attendance
# ------------------------------------------

average_attendance = df["Attendance"].mean()

print("Average Attendance:", round(average_attendance, 2))


# ==========================================
# 9. BAR CHART
# ==========================================

avg_marks = df.groupby("Department")["Marks"].mean()

plt.figure(figsize=(8, 5))

avg_marks.plot(kind="bar")

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()


# ==========================================
# 10. SCATTER PLOT
# ==========================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Attendance"],
    df["Marks"]
)

plt.title("Attendance vs Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")

plt.grid(True)

plt.show()


# ==========================================
# 11. HEATMAP
# ==========================================

numeric_columns = [
    "Student_ID",
    "Age",
    "Marks",
    "Attendance"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Data Correlation Heatmap")

plt.show()


# ==========================================
# 12. INSIGHTS
# ==========================================

print("\n========== INSIGHTS ==========")

print(
    "1. The average age of the students is",
    round(average_age, 2)
)

print(
    "2. The average marks of the students is",
    round(average_marks, 2)
)

print(
    "3. The average attendance of the students is",
    round(average_attendance, 2)
)

print(
    "4. The bar chart compares average marks across departments."
)

print(
    "5. The scatter plot shows the relationship between attendance and marks."
)

print(
    "6. The heatmap shows the correlation between numerical variables."
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# -------------------------------
# STEP 1: Upload CSV
# -------------------------------
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

# Load CSV
df = pd.read_csv(file_name)

print("Dataset loaded successfully!\n")

# -------------------------------
# STEP 2: Display Dataset
# -------------------------------
print("First 5 rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

# -------------------------------
# STEP 3: Basic Analysis
# -------------------------------
print("\nBasic Statistics:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

# -------------------------------
# STEP 4: Average
# -------------------------------
# Change this column name according to your CSV
numeric_columns = df.select_dtypes(include="number").columns

if len(numeric_columns) > 0:
    selected_column = numeric_columns[0]
    average = df[selected_column].mean()

    print("\nSelected Column:", selected_column)
    print("Average:", average)

# -------------------------------
# STEP 5: Bar Chart
# -------------------------------
# Automatically select categorical and numerical columns
categorical_columns = df.select_dtypes(include="object").columns

if len(categorical_columns) > 0 and len(numeric_columns) > 0:

    category_column = categorical_columns[0]
    value_column = numeric_columns[0]

    bar_data = df.groupby(category_column)[value_column].mean()

    plt.figure(figsize=(8, 5))
    bar_data.plot(kind="bar")

    plt.title("Average Value by Category")
    plt.xlabel(category_column)
    plt.ylabel("Average " + value_column)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# -------------------------------
# STEP 6: Scatter Plot
# -------------------------------
if len(numeric_columns) >= 2:

    x_column = numeric_columns[0]
    y_column = numeric_columns[1]

    plt.figure(figsize=(8, 5))
    plt.scatter(df[x_column], df[y_column])

    plt.title(x_column + " vs " + y_column)
    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.grid(True)

    plt.show()

# -------------------------------
# STEP 7: Heatmap
# -------------------------------
if len(numeric_columns) >= 2:

    correlation = df[numeric_columns].corr()

    plt.figure(figsize=(10, 6))

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Heatmap")
    plt.show()

# -------------------------------
# STEP 8: Insights
# -------------------------------
print("\n========== INSIGHTS ==========")

if len(numeric_columns) > 0:
    print(
        f"1. The average {numeric_columns[0]} is "
        f"{df[numeric_columns[0]].mean():.2f}."
    )

if len(numeric_columns) >= 2:
    corr_value = df[numeric_columns[0]].corr(df[numeric_columns[1]])

    print(
        f"2. The correlation between {numeric_columns[0]} and "
        f"{numeric_columns[1]} is {corr_value:.2f}."
    )

    if corr_value > 0.5:
        print("   This indicates a positive relationship.")
    elif corr_value < -0.5:
        print("   This indicates a negative relationship.")
    else:
        print("   This indicates a weak relationship.")

print("3. The bar chart helps compare average values across categories.")
print("4. The scatter plot helps identify relationships between numerical variables.")
print("5. The heatmap shows correlations among numerical variables.")
```
