```python
import numpy as np
```

```python
rows1 = int(input("Enter number of rows for Matrix A: "))
cols1 = int(input("Enter number of columns for Matrix A: "))

A = np.zeros((rows1, cols1))

print("Enter elements of Matrix A:")

for i in range(rows1):
    for j in range(cols1):
        A[i][j] = float(input(f"A[{i+1}][{j+1}]: "))

print("\nMatrix A:")
print(A)
```

```python
rows2 = int(input("Enter number of rows for Matrix B: "))
cols2 = int(input("Enter number of columns for Matrix B: "))

B = np.zeros((rows2, cols2))

print("Enter elements of Matrix B:")

for i in range(rows2):
    for j in range(cols2):
        B[i][j] = float(input(f"B[{i+1}][{j+1}]: "))

print("\nMatrix B:")
print(B)
```

```python
if A.shape == B.shape:
    addition = A + B

    print("Matrix A:")
    print(A)

    print("\nMatrix B:")
    print(B)

    print("\nA + B:")
    print(addition)
else:
    print("Error: Both matrices must have the same dimensions.")
```

```python
if A.shape == B.shape:
    subtraction = A - B

    print("Matrix A:")
    print(A)

    print("\nMatrix B:")
    print(B)

    print("\nA - B:")
    print(subtraction)
else:
    print("Error: Both matrices must have the same dimensions.")
```

```python
if cols1 == rows2:
    multiplication = np.matmul(A, B)

    print("Matrix A:")
    print(A)

    print("\nMatrix B:")
    print(B)

    print("\nA × B:")
    print(multiplication)
else:
    print("Error: Columns of Matrix A must equal rows of Matrix B.")
```

```python
transpose_A = A.T

print("Matrix A:")
print(A)

print("\nTranspose of A:")
print(transpose_A)
```

```python
transpose_B = B.T

print("Matrix B:")
print(B)

print("\nTranspose of B:")
print(transpose_B)
```

```python
if rows1 == cols1:
    determinant_A = np.linalg.det(A)

    print("Matrix A:")
    print(A)

    print("\nDeterminant of A:", round(determinant_A, 2))
else:
    print("Error: Determinant can only be calculated for a square matrix.")
```

```python
if rows2 == cols2:
    determinant_B = np.linalg.det(B)

    print("Matrix B:")
    print(B)

    print("\nDeterminant of B:", round(determinant_B, 2))
else:
    print("Error: Determinant can only be calculated for a square matrix.")
```
